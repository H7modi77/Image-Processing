```python
# Step#4 (c): Convert to YUV and separate channels
import cv2

img = cv2.imread('images/input.jpg')

# Convert to YUV
yuv_img = cv2.cvtColor(img, cv2.COLOR_BGR2YUV)

# Display the channels
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(yuv_img[:, :, 0], cmap='gray')
axes[0].set_title('Y channel (Luminance)')
axes[0].axis('off')

axes[1].imshow(yuv_img[:, :, 1], cmap='gray')
axes[1].set_title('U channel (Blue projection)')
axes[1].axis('off')

axes[2].imshow(yuv_img[:, :, 2], cmap='gray')
axes[2].set_title('V channel (Red projection)')
axes[2].axis('off')

plt.tight_layout()
plt.show()

# Save the channels
cv2.imwrite('images/Y_channel.jpg', yuv_img[:, :, 0])
cv2.imwrite('images/U_channel.jpg', yuv_img[:, :, 1])
cv2.imwrite('images/V_channel.jpg', yuv_img[:, :, 2])
print("✅ YUV channels saved")
```

## Task (Geometric Transformations) — 1) Resize

```python
# Task - 1) Increase the size of the image
import cv2

img = cv2.imread('images/input.jpg')
h, w = img.shape[:2]

scale = 1.5   # 150% larger
resized = cv2.resize(img, None, fx=scale, fy=scale, interpolation=cv2.INTER_LINEAR)

# Display
resized_rgb = cv2.cvtColor(resized, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(6, 4))
plt.imshow(resized_rgb)
plt.title(f'Resized {scale}x — New size: {resized.shape[1]}x{resized.shape[0]}')
plt.axis('off')
plt.show()

cv2.imwrite('images/resized.jpg', resized)
print("Original size:", (w, h))
print("New size:", (resized.shape[1], resized.shape[0]))
```

## # Task - 2) Rotate the image by 120 degrees


```python
# Task - 2) Rotate the image by 120 degrees
import cv2

img = cv2.imread('images/input.jpg')
h, w = img.shape[:2]
center = (w // 2, h // 2)

M_rot = cv2.getRotationMatrix2D(center, 120, 1.0)   # 120°, scale = 1
rotated = cv2.warpAffine(img, M_rot, (w, h))

rotated_rgb = cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(6, 4))
plt.imshow(rotated_rgb)
plt.title('Rotated 120°')
plt.axis('off')
plt.show()

cv2.imwrite('images/rotated_120.jpg', rotated)
print("✅ Image rotated by 120 degrees")
```

## Task (Geometric Transformations) — 3) Sheer

```python
# Task - 3) Perform Sheer operations over the image
import cv2
import numpy as np

img = cv2.imread('images/input.jpg')
h, w = img.shape[:2]

# Horizontal shear (x) by 0.5
M_shear_x = np.float32([[1, 0.5, 0],
                        [0, 1,   0]])
shear_x = cv2.warpAffine(img, M_shear_x, (int(w * 1.5), h))

# Vertical shear (y) by 0.5
M_shear_y = np.float32([[1,   0, 0],
                        [0.5, 1, 0]])
shear_y = cv2.warpAffine(img, M_shear_y, (w, int(h * 1.5)))

# Display
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(cv2.cvtColor(shear_x, cv2.COLOR_BGR2RGB))
axes[0].set_title('Shear X (horizontal)')
axes[0].axis('off')

axes[1].imshow(cv2.cvtColor(shear_y, cv2.COLOR_BGR2RGB))
axes[1].set_title('Shear Y (vertical)')
axes[1].axis('off')

plt.tight_layout()
plt.show()

cv2.imwrite('images/shear_x.jpg', shear_x)
cv2.imwrite('images/shear_y.jpg', shear_y)
print("✅ Horizontal and vertical shear applied")
```

## Task2 — 1) Negative Image

```python
# Task2 - 1) Create a Negative Image
import cv2

img = cv2.imread('images/input.jpg', cv2.IMREAD_GRAYSCALE)

negative = 255 - img

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(img, cmap='gray'); axes[0].set_title('Original'); axes[0].axis('off')
axes[1].imshow(negative, cmap='gray'); axes[1].set_title('Negative'); axes[1].axis('off')
plt.tight_layout()
plt.show()

cv2.imwrite('images/negative.jpg', negative)
print("✅ Negative image created")
```

## Task2 — 2) Log Transformation

```python
# Task2 - 2) Log Transformation
import cv2
import numpy as np

img = cv2.imread('images/input.jpg', cv2.IMREAD_GRAYSCALE)

# c = 255 / log(1 + max_pixel)
c = 255 / np.log(1 + np.max(img))

log_img = c * np.log(1 + img.astype(np.float32))
log_img = np.array(log_img, dtype=np.uint8)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(img, cmap='gray'); axes[0].set_title('Original'); axes[0].axis('off')
axes[1].imshow(log_img, cmap='gray'); axes[1].set_title('Log Transformation'); axes[1].axis('off')
plt.tight_layout()
plt.show()

cv2.imwrite('images/log_transform.jpg', log_img)
print("✅ Log transformation applied, c =", c)
```

## # Task2 - 3) Power Law (Gamma) - gamma that increases contrast


```python
# Task2 - 3) Power Law (Gamma) - gamma that increases contrast
import cv2
import numpy as np

img = cv2.imread('images/input.jpg', cv2.IMREAD_GRAYSCALE)

gamma = 0.4   # value that increases contrast in dark regions

power_img = np.power(img / 255.0, gamma)
power_img = np.array(power_img * 255, dtype=np.uint8)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(img, cmap='gray'); axes[0].set_title('Original'); axes[0].axis('off')
axes[1].imshow(power_img, cmap='gray'); axes[1].set_title(f'Power Law (γ={gamma})'); axes[1].axis('off')
plt.tight_layout()
plt.show()

cv2.imwrite('images/power_law.jpg', power_img)
print("✅ Power law applied with gamma =", gamma)
```
