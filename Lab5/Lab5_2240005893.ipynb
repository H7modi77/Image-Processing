## Imports


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

## Read Image


```python
# Read the image
image = cv2.imread('images/Parrot.png')

if image is None:
    raise FileNotFoundError("Put Parrot.png inside the images folder.")

# Convert from BGR to RGB
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 4))
plt.imshow(image)
plt.title('Original Image')
plt.axis('off')
plt.show()
```

# Task 1 
## 7×7 Box Filter

Convolve the image with a 7×7 box filter and display the original and smoothed images side by side.

```python
# Apply a 7x7 box filter
smoothed_box = cv2.blur(image, (7, 7))

# Display original and smoothed images side by side
fig, ax = plt.subplots(1, 2, figsize=(12, 5))

ax[0].imshow(image)
ax[0].set_title('Original Image')
ax[0].axis('off')

ax[1].imshow(smoothed_box)
ax[1].set_title('Smoothed Image - 7x7 Box Filter')
ax[1].axis('off')

plt.tight_layout()
plt.show()
```

# Task 2 
## Gaussian Filtering

Apply a 5×5 Gaussian filter and then a 21×21 Gaussian filter.Display the original and both filtered images side by side.

```python
# Apply 5x5 Gaussian filter
gaussian_5 = cv2.GaussianBlur(image, (5, 5), sigmaX=0)

# Apply 21x21 Gaussian filter
gaussian_21 = cv2.GaussianBlur(image, (21, 21), sigmaX=0)

# Display the three images side by side
fig, ax = plt.subplots(1, 3, figsize=(15, 5))

ax[0].imshow(image)
ax[0].set_title('Original Image')
ax[0].axis('off')

ax[1].imshow(gaussian_5)
ax[1].set_title('Gaussian Filter 5x5')
ax[1].axis('off')

ax[2].imshow(gaussian_21)
ax[2].set_title('Gaussian Filter 21x21')
ax[2].axis('off')

plt.tight_layout()
plt.show()
```

# Task 3 
## 3×3 Laplacian Sharpening

Steps from the lab manual:
1. Read the image.
2. Convert the image to `float32`.
3. Use `cv2.Laplacian()`.
4. Sharpen using `np.clip(image_float - laplacian, 0, 1)`.
5. Plot the original, Laplacian result, and sharpened image.

```python
# Convert image to float32 and normalize to [0, 1]
image_float = image.astype(np.float32) / 255.0

# Apply a 3x3 Laplacian filter
laplacian = cv2.Laplacian(image_float, cv2.CV_32F, ksize=3)

# Sharpen the image
sharpened_laplacian = np.clip(image_float - laplacian, 0, 1)

# Display original, Laplacian result, and sharpened image
fig, ax = plt.subplots(1, 3, figsize=(15, 5))

ax[0].imshow(image)
ax[0].set_title('Original Image')
ax[0].axis('off')

ax[1].imshow(np.clip(laplacian, 0, 1))
ax[1].set_title('Laplacian Filter 3x3')
ax[1].axis('off')

ax[2].imshow(sharpened_laplacian)
ax[2].set_title('Sharpened Image')
ax[2].axis('off')

plt.tight_layout()
plt.show()
```
